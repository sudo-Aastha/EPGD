### Find Unique Vowel-Only Substrings

First, let's define the input string and a helper function to determine if a character is a vowel.

```python
input_string = "aeiouaeiouhelloaeiouworld"
vowels = "aeiouAEIOU"

def is_vowel(char):
    return char in vowels

print(f"Input string: {input_string}")
```

Now, we'll iterate through all possible substrings of the `input_string`. For each substring, we'll check if all its characters are vowels. We'll use a `set` to store only the unique vowel-only substrings.

```python
unique_vowel_substrings = set()
n = len(input_string)

for i in range(n):
    for j in range(i, n):
        substring = input_string[i : j + 1]

        # Check if all characters in the substring are vowels
        is_all_vowels = True
        for char in substring:
            if not is_vowel(char):
                is_all_vowels = False
                break

        if is_all_vowels:
            unique_vowel_substrings.add(substring)

print(f"Unique vowel-only substrings found: {sorted(list(unique_vowel_substrings))}")
```

### Debugging a Recursive Fibonacci Function

Recursive Fibonacci functions are elegant but can be problematic with certain inputs or for large numbers due to performance and recursion depth limits. We'll define a basic recursive Fibonacci function first.

```python
def fibonacci_recursive(n):
    """Calculates the nth Fibonacci number recursively."""
    if n <= 0:
        return 0
    elif n == 1:
        return 1
    else:
        return fibonacci_recursive(n - 1) + fibonacci_recursive(n - 2)

print("Original recursive Fibonacci function defined.")
```

Now, let's add a wrapper function or modify the existing one to include debugging for input constraints, specifically handling non-integer and negative inputs. We'll also consider a practical limit for large `n` to avoid excessive computation or recursion depth errors, though the core issue with recursion depth for very large `n` would require memoization or iteration.

### Converting to an Object-Oriented Solution: Fibonacci Calculator

To improve scalability and encapsulate the Fibonacci logic, we can define a `FibonacciCalculator` class. This class will handle the recursive calculation, incorporate memoization to optimize performance for larger numbers, and include robust input validation.

```python
class FibonacciCalculator:
    def __init__(self):
        # Initialize a dictionary for memoization (caching results)
        self._memo = {}

    def calculate(self, n):
        """Calculates the nth Fibonacci number with input validation and memoization."""
        # Input Validation (similar to debugged_fibonacci)
        if not isinstance(n, int):
            print(f"DEBUG ERROR: Input must be an integer, but received type {type(n)}.")
            return None
        if n < 0:
            print(f"DEBUG ERROR: Input cannot be negative, but received {n}.")
            return None

        # Performance warning for very large numbers if not using memoization correctly
        # With memoization, this warning is less critical, but still good to note for extreme cases
        if n > 1000: # A higher practical limit with memoization
            print(f"WARNING: Calculating fibonacci({n}) might still be slow due to recursion depth or memory for very large 'n'.")

        # Check if the result is already memoized
        if n in self._memo:
            return self._memo[n]

        # Base cases
        if n == 0:
            result = 0
        elif n == 1:
            result = 1
        else:
            # Recursive step with memoization
            result = self.calculate(n - 1) + self.calculate(n - 2)

        # Store the result in memo for future use
        self._memo[n] = result
        return result

print("FibonacciCalculator class defined.")
```

Now, let's create an instance of our `FibonacciCalculator` and test it with various inputs, including those that previously caused issues or warnings.

```python
print("--- Testing Object-Oriented Fibonacci Calculator ---")

fib_calc = FibonacciCalculator()

# Test valid inputs
print(f"Fibonacci(5): {fib_calc.calculate(5)}")
print(f"Fibonacci(0): {fib_calc.calculate(0)}")
print(f"Fibonacci(1): {fib_calc.calculate(1)}")
print(f"Fibonacci(10): {fib_calc.calculate(10)}")

# Test invalid inputs
print(f"Fibonacci(-3): {fib_calc.calculate(-3)}")
print(f"Fibonacci(4.5): {fib_calc.calculate(4.5)}")
print(f"Fibonacci('xyz'): {fib_calc.calculate('xyz')}")

# Test a larger input to show memoization benefits (without hitting recursion limit for simple demo)
print(f"Fibonacci(35): {fib_calc.calculate(35)}")
print(f"Fibonacci(50): {fib_calc.calculate(50)}") # This would be extremely slow without memoization
```

### Resource Allocation: Task Scheduling Problem

**Problem:** You have a list of tasks, each with a specific processing time, and a fixed number of servers. Your goal is to assign all tasks to these servers such that the maximum total processing time (load) on any single server is minimized. This is a common load balancing problem.

**Chain of Thought for a Greedy Procedural Solution:**

1.  **Understand the Goal:** Minimize the time when the last task finishes across all servers. This means we want to keep server loads as balanced as possible.
2.  **Input:** A list of task processing times (e.g., `[10, 5, 8, 3, 12, 7]`) and the number of available servers (e.g., `3`).
3.  **Strategy:** A simple and effective greedy heuristic is to sort the tasks in descending order of processing time and then assign each task, one by one, to the server that currently has the *least* total load. This helps prevent large tasks from disproportionately loading a single server later on.
4.  **Initialization:** Start with all servers having a load of 0.
5.  **Assignment:** For each task (from largest to smallest), find the server with the minimum current load and add the task's processing time to that server's load.
6.  **Result:** After all tasks are assigned, the maximum value among the server loads will be our minimized maximum completion time.

```python
def schedule_tasks_procedural(tasks, num_servers):
    """Assigns tasks to servers using a greedy approach to minimize max load."""
    if not tasks or num_servers <= 0:
        print("Error: Tasks list cannot be empty and num_servers must be positive.")
        return [], 0

    # Initialize server loads (each server starts with 0 load)
    server_loads = [0] * num_servers

    # Strategy: Sort tasks in descending order for better load balancing with greedy approach
    sorted_tasks = sorted(tasks, reverse=True)

    # Assign each task to the server with the current minimum load
    for task_time in sorted_tasks:
        # Find the server with the minimum current load
        min_load_server_index = server_loads.index(min(server_loads))

        # Assign the task to that server
        server_loads[min_load_server_index] += task_time

    max_completion_time = max(server_loads)
    return server_loads, max_completion_time

# --- Test the procedural solution ---
tasks_procedural = [10, 5, 8, 3, 12, 7, 4, 6, 9]
num_servers_procedural = 3

print(f"\n--- Procedural Task Scheduling ---")
print(f"Tasks: {tasks_procedural}")
print(f"Number of Servers: {num_servers_procedural}")

final_loads_procedural, max_time_procedural = schedule_tasks_procedural(tasks_procedural, num_servers_procedural)

print(f"Final Server Loads: {final_loads_procedural}")
print(f"Maximum Completion Time: {max_time_procedural}")
```

### Converting to an Object-Oriented Solution

The procedural solution works, but an object-oriented approach can offer several advantages, especially for larger, more complex systems:

*   **Encapsulation:** Grouping data (like server load, tasks assigned) and behavior (adding tasks, getting load) into objects (`Server`, `TaskScheduler`).
*   **Modularity:** Easier to understand, test, and maintain individual components.
*   **Reusability:** `Server` objects could be reused in different scheduling algorithms or simulations.
*   **Extensibility:** Easier to add new features (e.g., server capacity, task priorities) without rewriting core logic.

We'll create two main classes:
1.  `Server`: Represents an individual server, managing its load.
2.  `TaskScheduler`: Manages a collection of `Server` objects and implements the scheduling logic.

```python
class Server:
    def __init__(self, server_id):
        self.server_id = server_id
        self.current_load = 0
        self.assigned_tasks = [] # Optional: keep track of assigned tasks

    def add_task(self, task_time):
        """Adds a task's processing time to the server's load."""
        self.current_load += task_time
        self.assigned_tasks.append(task_time)

    def __repr__(self):
        return f"Server(ID={self.server_id}, Load={self.current_load}, Tasks={self.assigned_tasks})"

class TaskScheduler:
    def __init__(self, num_servers):
        if num_servers <= 0:
            raise ValueError("Number of servers must be positive.")
        self.servers = [Server(i) for i in range(num_servers)]

    def schedule_tasks(self, tasks):
        """Schedules tasks to servers using the greedy approach."""
        if not tasks:
            print("No tasks to schedule.")
            return

        # Sort tasks in descending order
        sorted_tasks = sorted(tasks, reverse=True)

        for task_time in sorted_tasks:
            # Find the server with the minimum current load
            # We can use min with a custom key for Server objects
            min_load_server = min(self.servers, key=lambda server: server.current_load)

            # Assign the task to that server
            min_load_server.add_task(task_time)

    def get_server_loads(self):
        """Returns a list of current loads for all servers."""
        return [server.current_load for server in self.servers]

    def get_max_completion_time(self):
        """Returns the maximum completion time among all servers."""
        if not self.servers:
            return 0
        return max(self.get_server_loads())

    def __repr__(self):
        return f"TaskScheduler(Servers={self.servers})"

# --- Test the object-oriented solution ---
tasks_oo = [10, 5, 8, 3, 12, 7, 4, 6, 9]
num_servers_oo = 3

print(f"\n--- Object-Oriented Task Scheduling ---")
print(f"Tasks: {tasks_oo}")
print(f"Number of Servers: {num_servers_oo}")

scheduler = TaskScheduler(num_servers_oo)
scheduler.schedule_tasks(tasks_oo)

print(f"Final Server States: {scheduler.servers}")
print(f"Maximum Completion Time: {scheduler.get_max_completion_time()}")
```
