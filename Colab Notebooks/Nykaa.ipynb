```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

# 1. Load the dataset directly from a public source
df = pd.read_csv("/content/sample_data/nykaa.csv")

df.head()
```

```python
# 2. Filter for the specified brands
target_brands = ["Biotique", "Himalaya", "Lotus Herbals"]
df_filtered = df[df["Product Brand"].isin(target_brands)]

```

```python
plt.figure(figsize=(8, 5))
sns.countplot(data=df_filtered, x="Product Brand", order=target_brands)

# Add titles and labels
plt.title("Number of Products per Brand")
plt.xlabel("Product Brand")
plt.ylabel("Count")

# Show the plot
plt.show()
```

```python
# Plot 1: Mean Product Prices
plt.figure(figsize=(8, 5))
sns.barplot(
    data=df_filtered, x="Product Brand", y="Product Price", estimator="mean", errorbar=None
)
plt.title("Mean Product Prices by Brand")
plt.xlabel("Brand")
plt.ylabel("Mean Price")

plt.show()

# Plot 2: Median Product Prices
plt.figure(figsize=(8, 5))
sns.barplot(
    data=df_filtered, x="Product Brand", y="Product Price", estimator="median", errorbar=None
)
plt.title("Median Product Prices by Brand")
plt.xlabel("Product Brand")
plt.ylabel("Median Price")
plt.show()
```

```python
# 3. Plot the stacked histogram with 10 bins
plt.figure(figsize=(10, 6))
sns.histplot(
    data=df_filtered,
    x="Product Price",
    hue="Product Brand",
    multiple="stack",
    bins=10,
    edgecolor="white",
)

plt.title("Product Price Distribution by Brand")
plt.xlabel("Product Price")
plt.ylabel("Count")
plt.show()

```

```python


# Customize whiskers to extend to the 5th and 95th percentiles

sns.boxplot(data=df_filtered, x="Product Brand", y="Product Rating")
plt.title("Product Price Distribution by Brand")
plt.xlabel("Product Brand")
plt.ylabel("Product Rating")
plt.show()
```

```python
selected_columns = [
    "Product Price",
    "Product Rating",
    "Product Reviews Count",
]
correlation_matrix = df_filtered[selected_columns].corr()

plt.figure(figsize=(8, 6))
sns.heatmap(correlation_matrix, annot=True, cmap="coolwarm", fmt=".2f")
plt.title("Correlation Heatmap of Product Price, Rating, and Reviews Count")
plt.show()
```
