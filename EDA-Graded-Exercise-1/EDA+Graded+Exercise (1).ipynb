You should categorise the dates into seasons in “Dress Sales” data to simplify the analysis according to the following criteria:
- June, July and August: Summer.
- September, October and November: Autumn.
- December, January and February: WInter.
- March, April and May: Spring.




```python
# Categorizing the remaining date columns into seasons

# Summer: June, July, August
summer_cols = ['09-06-2013', '31-08-2013', '29-08-2013', '09-08-2013', '10-06-2013']
inp1['Summer'] = inp1[inp1.columns.intersection(summer_cols)].sum(axis=1)

# Autumn: September, October, November
autumn_cols = ['09-10-2013', '14-09-2013', '16-09-2013', '18-09-2013', '20-09-2013', '22-09-2013', '24-09-2013', '28-09-2013']
inp1['Autumn'] = inp1[inp1.columns.intersection(autumn_cols)].sum(axis=1)

# Winter: December, January, February
winter_cols = ['09-02-2013', '09-12-2013', '10-12-2013']
inp1['Winter'] = inp1[inp1.columns.intersection(winter_cols)].sum(axis=1)

# Spring: March, April, May
spring_cols = ['09-04-2013']
inp1['Spring'] = inp1[inp1.columns.intersection(spring_cols)].sum(axis=1)

# Display the seasonal summary
display(inp1[['Dress_ID', 'Summer', 'Autumn', 'Winter', 'Spring']].head())
```

You can see that there are two types of variables one with a large number of missing values and another is very less number of missing values. These two columns can be categorized as:

Type-1: Missing values are very less (around 2 or 3 missing values): Price, Season, NeckLine, SleeveLength, Winter and Autumn.

### Multivariate analysis

Which of the following pair of “Style” and “Price” category has the highest average of positive recommendations?
- Price: medium and style: vintage
- Price: medium and style: cute
- Price: very high and style: party
- Price: low and style: sexy


```python
# Calculate the pivot table for the heatmap
style_price_rec = inp0.pivot_table(index='Style', columns='Price', values='Recommendation', aggfunc='mean')

# Plot the heatmap
plt.figure(figsize=(10, 8))
sns.heatmap(style_price_rec, annot=True, cmap='RdYlGn')
plt.title('Heatmap of Recommendation Rate by Style and Price')
plt.show()

# Identify the highest average
max_val = style_price_rec.max().max()
print(f"The highest average recommendation rate in the heatmap is: {max_val}")
```

### Analysis of Style and Price Recommendation Rates
Based on the heatmap generated:
- **Cute / Medium**: 1.0 (Highest)
- **Party / Very-high**: ~0.82
- **Sexy / Low**: 0.75
- **Vintage / Medium**: 0.5

The pair with the highest average positive recommendation is **Price: medium and style: cute**.

Which of the following material type has no recommendation in summer and winter seasons?
- Mix and Milksilk
- Nylon and Rayon
- Microfiber and Silk
- Milksilk and Microfiber


```python
# Calculate the pivot table for Season, Material, and Recommendation
# We filter for only the main seasons to keep the heatmap clean
res = inp0.pivot_table(index='Material', columns='Season', values='Recommendation', aggfunc='mean')

# Plot the heatmap focusing on Summer and Winter
plt.figure(figsize=(12, 10))
sns.heatmap(res[['Summer', 'Winter']], annot=True, cmap='RdYlGn')
plt.title('Recommendation Rate by Material for Summer and Winter')
plt.show()

# Filtering the results to answer the specific question
zero_rec = res[(res['Summer'] == 0) & (res['Winter'] == 0)]
print("Materials with no recommendations in both Summer and Winter:")
print(zero_rec.index.tolist())
```
